```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
demographic = pd.read_stata("sec_a_households_demographic.dta")
electricity = pd.read_stata("sec_2_electricity.dta")
cooking = pd.read_stata("sec_3_cooking.dta")
lighting = pd.read_stata("sec_6_lighting.dta")
quantities = pd.read_stata("sec_7_quantities_used.dta")
```

```python
datasets = {
    "Demographic": demographic,
    "Electricity": electricity,
    "Cooking": cooking,
    "Lighting": lighting,
    "Quantities Used": quantities
}
```

```python
for name, df in datasets.items():
    print(name)
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("-" * 50)
```
